# California Housing Price Prediction
Reproducible linear-regression baseline using local housing data. The train/test split occurs before preprocessing to avoid fitting imputation or encoding on the test set.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from train import run
data_path = Path('housing.csv')
data = pd.read_csv(data_path)
data.head()
```

## Explore the data
The relationship below is descriptive, not causal.

```python
plt.scatter(data['median_income'], data['median_house_value'], alpha=0.3)
plt.xlabel('Median income')
plt.ylabel('Median house value (USD)')
plt.show()
```

## Train and evaluate
A fixed 80/20 split and preprocessing pipeline provide a reproducible baseline. R², MSE, RMSE and MAE are evaluated on the held-out test set.

```python
results = run(data_path)
results
```

## Limitations
This historical dataset is not a current property valuation service. A random split can overestimate geographic generalization; spatial splits and additional baselines are useful future work.
